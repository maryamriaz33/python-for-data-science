```python
# Mini Project — Student Performance Analyzer
import numpy as np

marks = np.array([
    [85, 78, 90, 88],
    [72, 80, 75, 79],
    [91, 88, 95, 92],
    [65, 70, 68, 72],
    [88, 92, 85, 90],
    [76, 82, 80, 78],
    [95, 90, 94, 96],
    [60, 65, 30, 68],
    [84, 79, 88, 85],
    [73, 75, 77, 80]
])

subjects = ["Python", "SQL", "Statistics", "Machine Learning"]

# 1. Shape
print("Shape:", marks.shape)

# 2. Average marks of every student
student_average = np.mean(marks, axis=1)
print("Average marks of every student:", student_average)

# 3. Average marks of every subject
subject_average = np.mean(marks, axis=0)
print("Average marks of every subject:", subject_average)

# 4. Highest mark in every subject
print("Highest marks of every subject:", np.max(marks, axis=0))

# 5. Lowest mark in every subject
print("Lowest marks of every subject:", np.min(marks, axis=0))

# 6. Overall class average
print("Overall Class Average:", np.mean(marks))

# 7. Student with highest total marks
student_totals = np.sum(marks, axis=1)
highest_student = np.argmax(student_totals)

print("Student with Highest Total Marks:",
      highest_student + 1)

# 8. Students with average >= 80
high_performers = np.where(student_average >= 80)[0] + 1
print("Students with Average >= 80:", high_performers)

# 9. Students with average < 70
low_performers = np.where(student_average < 70)[0] + 1
print("Students with Average < 70:", low_performers)

# 10. Students who passed
passed = student_average >= 50
print("Number of Students Passed:", np.sum(passed))

# 11. Students who failed any subject
failed_students = np.where(np.any(marks < 50, axis=1))[0] + 1
print("Students who Failed Any Subject:", failed_students)

# 12. Highest performing subject
highest_subject = np.argmax(subject_average)

print("Highest Performing Subject:",
      subjects[highest_subject])
```
