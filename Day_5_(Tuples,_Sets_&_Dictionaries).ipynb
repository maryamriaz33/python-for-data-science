```python
# Creating tuples
b = (8,9,67)
print(b)
print(type(b))
```

```python
names = ('maryam','ali','manaal','rafia','zunairah')
print(names)
```

```python
# Acessing an Element
print(names[3])
print(b[1])
```

```python
# Identify the Index Number

# By Using count()
print(names.count("ali"))

#By Using index()
print(names.index("maryam"))
```

```python
# Tuple Packing
f = 'tulip',98,'roses',100
print(f)
print(type(f))
```

```python
# Tuple unpacking
student = (1234,"Yasir",18)
id,name,age = student
print(id)
print(name)
print(age)
```

```python
# Swapping with unpacking
a = 200
b = 34
a,b = b,a
print(a)
print(b)
```

```python
# Creating a Sets
pets ={'dog','cat','parrot','horse','rabit'}
print(pets)
print(type(pets))
```

```python
# Add an element in set
pets.add("birds")
print(pets)
```

```python
# Remove an element
pets.remove("horse")
print(pets)
```

```python
# Discard an element
pets.discard('rabit')
print(pets)
```

```python
'''union()
intersection()
difference()
Remove duplicates'''

s = {56,25,32,22,92,90,22}
t = {1,3,5,2,1,22,67}

res1 = s.union(t)
res2 = s.intersection(t)
res3 = s.difference(t)
res4 = set(t)
print(res1)
print(res2)
print(res3)
print(res4)

```

```python
# Creating a Dictionaries
fruit = {1: 'mango',
         2:'apple',
         3:'banana'}
print(fruit)
```

```python
print(type(fruit))
```

```python
# Accessing Dict

# By using Get Function
print(fruit.get(2))

# By using the Key
print(fruit[1])
```

```python
# Adding Element in Dict
fruit[4] = 'Kiwi'
print(fruit)
```

```python
# Udpading in Dict
fruit[2]='Dargon'
print(fruit)
```

```python
fruit = {1: 'mango', 2: 'Dargon', 3: 'banana', 4: 'Kiwi'}
del fruit[3]
print(fruit)
```

```python
# Dict Methods
fruit.keys()
```

```python
fruit.values()
```

```python
fruit.items()
```

```python
# Loop Through Dictionaries
for key in fruit:
  print(key)
```

```python
for value in fruit.values():
  print(value)
```

```python
for key,value in fruit.items():
  print(key,':',value)
```

```python
if "apple" in fruit.values():
  print("The Fruit Exits!")
else:
  print("Invalid")
```

```python
# Dict with different Data Types
admin = {'name': 'amir',
         'admin_id': 15678,
         'email':'amirriaz34x@hotmail.com',
         'dep':'AI',
         'sal':45678.90,
         'skills':['python','java','prompt engineering'],
         'active':False
         }
for key,value in admin.items():
  print(key,':',value)
```

```python
# Nested Dict
students = {
    'student1':{
        'name': 'ali',
        'marks': 490
    },
    'student2':{
        'name':'raheel',
        'marks':'433'
    }
}
print(students)

```
